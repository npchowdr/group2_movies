# Q1 Interactive: Low-budget breakout finder (TH)

**Question:** Which low-budget films made far more than they cost, and do they have anything in common?

This is the interactive version of chart **1a** in `TH_exploratory_viz.ipynb`. Instead of fixing "low budget" at $9M and "breakout" at 6x, a studio can set its own **budget cap**, **genre**, and **breakout multiple** and see how often that kind of film pays off.

**How this notebook and the app fit together**
- Streamlit apps can't run inside a notebook. They have to be a `.py` file, so the page lives in **`q1_app.py`** and the chart logic lives in the group's **`apputil.py`** (the "Q1: low-budget breakouts" section).
- This notebook imports the same chart function from `apputil.py`, so the chart below is exactly what the app shows. It uses the group's data filter (2010-2018), so the numbers differ slightly from `TH_exploratory_viz.ipynb`.

````python
from apputil import load_data, breakout_figure, breakout_tag, breakout_summary

df = load_data("movie_roi.csv")
print(f"{len(df):,} films released 2010-2018")
````

## The chart
Hover over any dot to see the film. **Orange** = breakouts in the selection, **blue** = the rest of the selection, **gray** = every other film.

Default view: the notebook's original definitions (budget of $9M or less, gross of at least 6x budget).

````python
breakout_figure(df, max_budget=9e6, genre="All genres", multiple=6)
````

## Try other settings
In the app these are sliders. Here, change the three values and re-run the cell.

````python
MAX_BUDGET = 9e6      # budget cap in dollars
GENRE = "Horror"      # or "All genres", "Thriller", "Comedy", ...
MULTIPLE = 6          # breakout = worldwide gross of at least this many times the budget

s = breakout_summary(breakout_tag(df, MAX_BUDGET, GENRE, MULTIPLE))
print(f"{s['films']} films | {s['breakouts']} breakouts ({s['rate']:.0%}) | "
      f"all other films: {s['rest_rate']:.0%} | typical return {s['median_multiple']:.1f}x budget | "
      f"{s['lost_money']:.0%} lost money")
breakout_figure(df, MAX_BUDGET, GENRE, MULTIPLE)
````
